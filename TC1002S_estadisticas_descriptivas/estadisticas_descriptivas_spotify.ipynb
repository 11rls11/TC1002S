# Actividad Evaluable: Obtención de estadisticas descriptivas 

Importando las librerías necesarias para procesar los datos de Spotify y cargando datos (como el .csv esta en el mismo directorio que el código no tengo que poner toda la ruta) (W3Schools.com, s. f.):

```python
import pandas as pd

df = pd.read_csv("spotify.csv")
```

Verifico la cantidad de datos (no nulos, osea, con valor), las variables de cada vector (en este caso columnas) y que tipo de dato es cada columna (el Dtype) con el método info() (Encontrar el Tamaño de un Marco de Datos En Python, 16d. C.),

```python
df.info()
```

Para saber más de todos mis datos utilizaré el método *describe()* junto con el parametro *include = "all"*,  para generar un análisis de todas las estadísticas sin importar si son datos númericos o categoricos (datos que se pueden clasificar pero no tienen una jerarquía u orden matemático (Aprendizaje Automático de Datos Categóricos Con el Tutorial de Python, 2024)) y así llegar a conclusiones a partir de este informe de todas las columnas (GeeksforGeeks, 2025) y también determinaré la mediana de mis columnas númericas con el método *median()* y su parametro *numeric_only* como true, ya que para determinar una mediana el dato debe ser númerico (Malli, 2024).

```python
print(df.describe(include="all"))
```

```python
print(df.median(numeric_only=True))
```
